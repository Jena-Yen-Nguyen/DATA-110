**Problem 1: Confusion Matrix Metrics (Python as a Calculator)**

A binary classification model predicts whether rainfall will cause Flood (Yes/No).

Given:

	True Positives (TP) = 50
	False Positives (FP) = 8
	True Negatives (TN) = 32
	False Negatives (FN) = 10

Tasks:

	Calculate total predictions
	Compute:
	Accuracy
	Precision
	Recall
	F1 Score
Formula:

"Accuracy"=(TP+TN)/(TP+FP+TN+FN)

"Precision"=TP/(TP+FP)

"Recall"=TP/(TP+FN)

"F1"=2PR/(P+R)

```python
##CALCULATE TOTAL PREDICTIONS
#Initialize and assign variables
tp = 50
fp = 8
tn = 32
fn = 10
#Calculate total predictions
total = tp + fp+ tn+ fn
print("Total predictions: ",total)
```

```python
##COMPUTE ACCURACY
#Formula
a= (tp+tn)/total
#print
print("Accuracy: ", a)
```

```python
##COMPUTE PRECISION
#Formula
p= tp/(tp+fp)
#print
print("Precision: ", p)
```

```python
##COMPUTE RECALL
#Formula
r= tp/(tp+fn)
#print
print("Recall: ", r)
```

```python
##COMPUTE F1 SCORE
#Formula
F1 = (2*p*r)/(p+r)
#print
print("F1 score: ", F1)
```

**Problem 2: R² (Coefficient of Determination)**

In a regression model predicting river discharge, the following are known:

	Sum of Squares due to Regression (SSR) = 420
	Sum of Squares due to Error (SSE) = 180
Tasks

	Compute Total Sum of Squares (SST)
	Calculate R²
	Interpret whether the model explains more than 70% variance

Formula

SST=SSR+SSE

R^2=SSR/SST

```python
##COMPUTE SST
#variables
SSR = 420
SSE = 180
#Calculate SST
SST = SSR + SSE
print("Total Sum of Squares: ", SST)
```

```python
##CALCULATE R SQUARE
#Calculate R^2
R_square = SSR/SST
print("R^2: ", R_square)
```

```python
SSR = 420
SSE = 180
SST = SSR + SSE
R_square = SSR/SST

if R_square > 0.7:
  print("The model explains more than 70% variance")
elif R_square <= 0.7:
  print("The model explains less than or equal to 70% variance")
```

**Problem 3: Mean Squared Error (MSE)**

A model predicts temperature values.
The squared errors for 5 observations are: (2,4,1,9,4)

Tasks

	Store squared errors in Python
	Calculate Mean Squared Error (MSE)
	Print the result

Formula

MSE=(∑e^2)/n

```python
#Store SE
SE = (2,4,1,9,4)
#Calculate MSE
sum_SE = sum(SE)
n = len(SE)
MSE = sum_SE/n
print("MSE: ",MSE)
```

**Problem 4: Cook’s Distance(Influential Observation)**

For a regression model:

	Squared residual (eᵢ²) = 9
	Mean Squared Error (MSE) = 3
	Number of predictors (p) = 3
	Leverage value (hᵢᵢ) = 0.2
	Sample size (n) = 25
Tasks

	Compute Cook’s Distance
	Compare with threshold 4/n
	Decide whether the observation is influential

Formula

Di=((ei^2)/(p×MSE))×(hii/(1-hii)^2))

```python
##COMPUTE COOK'S DISTANCE
#Variables
ei = 9
MSE = 3
p = 3
hii = 0.2
#Calculate Cook's D
Di = (ei**2/(p*MSE))*(hii/(1-hii)**2)
print("Cook's Distance: ", Di)
```

```python
##INFLUENTIAL OBSERVATION
#Threshold
n = 25
threshold = 4/n
print("Threshold: ",threshold)
#Compare Di and threshold
if Di< threshold:
  print("Di < threshold")
  print("Observation is not influential")
else:
  print("Di > threshold")
  print("Observation is influential")

```
