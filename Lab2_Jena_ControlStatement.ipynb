```python
## Nested list
Main_list= [[1,2,3],[4,5,6],[7,8,9]]
```

```python
Combine_list = []
for sublist in Main_list:
  for ele in sublist:
    Combine_list.append(ele)
print(Combine_list)
```

```python
##print([ele outter loop to inner loop])
print([ele for sublist in Main_list for ele in sublist])
```
